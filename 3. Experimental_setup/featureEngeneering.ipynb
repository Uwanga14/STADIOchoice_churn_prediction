```python
import pandas
```

```python
df = pandas.read_csv('https://raw.githubusercontent.com/Uwanga14/netflix_customer_churn/refs/heads/main/preprocessed.csv')
```

```python
df.head(5)
```

```python
# 1. Combined viewing engagement
df['engagement_score'] = (
    df['watch_hours'] * df['avg_watch_time_per_day']
)
```

```python
df.head(5)
```

```python
# 2. Viewing intensity relative to inactivity
df['watch_intensity'] = (
    df['watch_hours'] / (df['last_login_days'] + 1)
).astype(int)
```

```python
df.head(5)
```

```python
df.watch_intensity.value_counts()
```

```python
# 3. Identify inactive customers
df['inactive_customer'] = (
    df['last_login_days'] >= 14
).astype(int)
```

```python
df.head(5)
```

0-Active
1-Inactive

```python
df.inactive_customer.value_counts()
```

```python
# 4. Identify accounts with multiple profiles
df['multiple_profiles'] = (
    df['number_of_profiles'] > 1
).astype(int)
```

```python
df.head(5)
```

```python
# Display the new features
df[
    [
        'watch_hours',
        'avg_watch_time_per_day',
        'last_login_days',
        'number_of_profiles',
        'engagement_score',
        'watch_intensity',
        'inactive_customer',
        'multiple_profiles'
    ]
].head()
```

```python
df.info()
```

```python
print(
    df.groupby('inactive_customer')['churned'].mean()
)
```

```python
#Recency Score
df['recency_score'] = 1 / (df['last_login_days'] + 1).round(2)
```

Customers who logged in recently are less likely to churn.

```python
df.head(5)
```

```python
print(
    df.groupby('subscription_type_Premium')['churned'].mean()
)
```

```python
#Premium Subscriber
df['premium_subscriber'] = (
df['subscription_type_Premium']
).astype(int)
```

```python
df.head(5)
```

```python
#Low Engagement
df['low_engagement'] = (
    df['watch_hours'] < 5
).astype(int)
```

```python
df.head(5)
```

```python
value_counts_low_engagement = df['low_engagement'].value_counts()
```

```python
value_counts_low_engagement
```

0- No
1- Yes
Accounts with low engagement are less than those who engage.
Very low viewing activity often indicates churn risk.

```python
#High Engagement
df['high_engagement'] = (
df['watch_hours'] > 20
).astype(int)
```

```python
df.head(5)
```

```python
value_counts_high_engagement = df['high_engagement'].value_counts()
value_counts_high_engagement
```

0- No
1- Yes
Highly engaged users are typically less likely to cancel.

```python
df['shared_account'] = (
    df['number_of_profiles'] >= 3
).astype(int)
```

```python
df.head(5)
```

Accounts shared by multiple users may be more valuable and less likely to churn.

```python
df['daily_engagement_ratio'] = (
df['avg_watch_time_per_day'] /
(df['watch_hours'] + 1)
).round(2)
```

```python
df.head(5)
```

```python
feature_corr = df[
    [
        'watch_hours',
        'avg_watch_time_per_day',
        'last_login_days',
        'number_of_profiles',
        'engagement_score',
        'watch_intensity',
        'inactive_customer',
        'multiple_profiles',
        'churned'
    ]
].corr()['churned'].sort_values(ascending=False)

print(feature_corr)
```

Customers who have not logged in recently and are classified as inactive show the highest likelihood of churn, with last_login_days (0.472) and inactive_customer (0.341) being the strongest positive predictors.

Higher platform engagement reduces churn risk, with watch_hours (-0.481), avg_watch_time_per_day (-0.218), and watch_intensity (-0.217) showing that more active viewers are less likely to cancel their subscriptions.
